**Accessing Substring from a String**

```python
#Positive Indexing
s = 'hello world'
print(s[0])
```

```python
#Negative Indexing - it will go from right to left , the last char will have index = -1, second last = -2
s = 'hello world'
print(s[-3])
```

```python
#Slicing - for extracting multiple chars from string
#Second value is always excluded even in negative slicing scenarios
s = 'helloworld'
print(s[0:5]) # 5 excluded so this print from 0 to 4
print(s[2:]) #for printing from 2 till last
print(s[:4]) #for printing from start till 3 because 4 will be excluded
print(s[0:11:2]) #third  value is for step size - that means if i have written 2 for step size then it will be printing second value from the current character

#giving negative third value
print(s[6:0:-1]) #for printing in reverse can also work as step size if given less than -1, 0th index will be excluded
print(s[6:0:-2])

#for reversing the string
print(s[::-1])
```

```python
s = 'hello world'
print(s[-5:]) #for printing world using negative indexing

#printing world in reverse
print(s[-1:-6:-1]) # here (s[-1:-6]) deosnt work beacause python moves in forward direction that +1 step size by default so you're asking it to move forward but stop somewhere behind — that's impossible, so it returns an empty string.
```

**Operations on Strings**

```python
#Arithmetic operation on string
print('delhi ' + 'mumbai')

print('delhi '*5)
```

```python
#relational operator

print('delhi' == 'mumbai')
print('delhi' == 'delhi')
print('delhi'!='Delhi')

#strings are compared lexicographically meaning the alphabet which comes earlier treated as smaller based on the ascii value
print('mumbai' > 'pune')
print('Pune' > 'pune') #P ASCII value smaller than p ASCII value
```

```python
#Logical Operators
#For strings empty string is false and non empty is true in case of and it will check both so later one will be printed
#In case of or if first one is non empty then no need to check the other one so first will be printed
print('hello' and 'world')

print('' and "akash") #false so empty string will be printed
print('akash' or '')

print(not '')
print(not 'akash')
```

**String Functions**

```python
#Common Functions - works of sets , tuples and dictionary also
# 1. len
# 2. max
# 3. min
# 4. sorted

print(len('hello'))

print(max('hello'))

print(sorted("hello")) #it returns a list of sorted chars based on ASCII value

#for sorting in descending order
print(sorted("hello",reverse=True))
```

**Functions specifically for Strings**

*   Capitalize
*   Title
*   Upper
*   Lower
*   Swapcase









```python
s = 'hello world'
print(s.capitalize()) #converts first character of sentence to capital

print(s.title()) #converts first char of every letter to capital

print(s.upper()) #converts whole string to capital

print(s.lower())

print(s.swapcase()) #converts lower to capital and capital to lower
```

**More Functions - Count, Find, Index**

```python
#count - count the frequency count of a substring in a string
s = "hello world"
print(s.count('l'))

#find - find the starting index of a substring in a string, if we find something that doesnt exists it gives -1
print(s.find('ld'))

#index - same as find , but throws error if we find something that doesnt exist
print(s.index('ld'))
```

**More Functions - endswith, startswith, format**

```python
#endswith
s = "my name is akash"
print(s.endswith('ak'))

#startswith
print(s.startswith('m'))

#format
a = 'akash'
b = 'male'

print('hy my name is {} and gender is {}'.format(a,b))

#we can also provide numbering in curly braces that is which number in format we want in string

print('hy my name is {1} and gender is {0}'.format(b,a))
```

**More Functions - isalnum, isalpha, isdigit, isidentifier**

```python
#isalnum
s = 'akash123'
print(s.isalnum()) #true

#isalpha
print(s.isalpha()) #this will be false becuase string is not alpha its alphanumeric

#isdigit
print(s.isdigit()) #it checks if only digits

#isidentifier
print(s.isidentifier()) #it checks for indetifier
```

**Split, Join, replace, strip**

```python
#split - split the string word by word, by default based on space , we can also provide on which we want to split, returns the list of splitted words/strings

s = "hi my name is akash"
print(s.split())

a = "akash kush"
print(a.split())

#join - it is the opposite of split , it joins the string to form a sentence from the list of chars or strings or joins the chars to forms a string
c = ['hi', 'my', 'name', 'is', 'akash']
print(" ".join(c))

#replace
d = 'my name is akash'
print(d.replace('akash','ankur'))

#strip - it removes all the trailing spaces
e = "     akash   sndos          "
print(e.strip())
```
