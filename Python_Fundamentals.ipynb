```python
# write code for addition of two numbers

#take input of two numbers
a = int(input('enter first number '))
b = int(input('enter second number '))
# print(type(a),type(b))
# print(a,b)

#add two variables , by default python takes input values as string so we have to convert them to integer explicitly
# Important Note :- pythons type conversion does not the actual value it creates another value or object in the memory of the type we write for conversion or we can
# change the type directly while taking the input in this way the value will be stored as integer permanently because we stored it in the variable as integer
c = int(a)+int(b)

#print the result
print(c)

print(type(c)) #this is int now

print(type(a)) #type of a still be string because it is not changed new value in memory is formed after type conversion
```

**Type Conversion**

```python
#Implicit vs Explicit Type Conversion

print(5+5.6) #this is implicit type conversion

#but implicit type conversion doesnt works always so we have to explicitly change the types
# we use different methods for that like for int we do int(), float() etc.

print(int(5.67)) # convert it to integer from float
print(int("5"))
```
