```python
#Arithmetic Operator
# +, -, *, /, // - Integer division(it will return integer if answer is in points)

print(5//2) # answer will be 2

# ** power of operator
print(5**2) # ans is 25
```

```python
#Logical Operator
# and, or, not
print(1 and 0)
print(1 or 0)
print(not 1)
```

```python
#Bitwise Operator
# &, |

#Bitwise and
print(2 & 3)

#Bitwise or
print(2 | 3)

#Bitwise xor
print(2 ^ 3)

#bitwise not ~  here we make 0 to 1 and 1 to 0
print(~3)

#leftshift
print(5 << 2)

#right shift
print(5 >> 2)

#assignment operator =
a = 2 # = is assignment operator

#Membership Operator (in/not in)

print('D' in 'Delhi')  #output will be true

print('D' not in 'delhi') # output will be true

print(1 in [2,3,4,5]) # false
```

```python
#write a program for addition of this series 1/1! + 2/2! + 3/3! + 4/4! ....... n
n = int(input("enter the number "))
fact = 1
sum = 0

for i in range(1,n+1):
  fact = fact * i
  sum = sum + i/fact

print(sum)
```

**Nested Loops**

```python
for i in range(1,5):
  for j in range(1,5):
    print(i,j)
```

```python
# print pattern
# *
# **
# ***
# ****

for i in range(1,5):
  for j in range(1,i+1):
    print('*',end='')
  print()
```

```python
#pattern
# 1
# 121
# 12321
# 1234321

for i in range(1,5):
  for j in range(1,i+1):
    print(j,end='')
  for k in range(i-1,0,-1):
    print(k,end='')
  print()
```

```python
#Loop Control Statement
# Break
# Continue
# Pass

#Program to find prime numbers between two numbers

a = int(input('lower range'))
b = int(input('upper range'))


for i in range(a,b+1):
  for j in range(2,i):
    if i%j==0:
      break
  else:
    print(i)
```

```python
#continue Statement

for i in range(1,10):
  if i==5:
    continue
  print(i)


#pass statement - used for passing a code to escape the error

for i in range(1,10):
  pass     # now it will not throw an error
```
